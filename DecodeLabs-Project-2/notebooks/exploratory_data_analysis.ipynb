# DecodeLabs Project 2 — Exploratory Data Analysis (EDA)

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```
df = pd.read_excel("../data/raw/dataset.xlsx")
df.head()
```

```
print("Shape:", df.shape)
df.info()
df.isna().sum()
```

## Basic Statistics

```
numeric = df.select_dtypes(include=np.number)
numeric.describe().T
```

```
pd.DataFrame({"count": numeric.count(), "mean": numeric.mean(), "median": numeric.median()})
```

## Outlier Detection (1.5 × IQR)

```
for col in numeric.columns:
    s=numeric[col].dropna(); q1=s.quantile(.25); q3=s.quantile(.75); iqr=q3-q1
    lo=q1-1.5*iqr; hi=q3+1.5*iqr
    print(col, "outliers:", int(((s<lo)|(s>hi)).sum()))
```

## Distributions

```
for col in numeric.columns:
    plt.figure(figsize=(7,4)); plt.hist(numeric[col].dropna(), bins=20)
    plt.title(f"Distribution of {col}"); plt.xlabel(col); plt.ylabel("Frequency"); plt.show()
```
